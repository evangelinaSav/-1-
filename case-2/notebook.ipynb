```python
import pandas as pd
import numpy as np

data = pd.read_csv('/without_rules.csv')
```

```python
data.head()
```

```python
#друг А
friends_A_forA = data[data['Subscription'] == 'A'].sort_values(by='Videos Watched')
sum_of_A = friends_A_forA['Videos Watched'].values[-200:].sum()

friends_A_forB = data[data['Subscription'] == 'B'].sort_values(by='Videos Watched')
sum_of_B = friends_A_forB['Videos Watched'].values[:200].sum()

inxA_A = friends_A_forA['Videos Watched'].index[-200:]
inxA_B = friends_A_forB['Videos Watched'].index[:200]
inxA = inxA_A + inxA_B
```

```python
import matplotlib.pyplot as plt


plt.bar(['A', 'B'], [sum_of_A, sum_of_B])
plt.title('Вовлеченность')
plt.xlabel('Продукт')
plt.ylabel('Количество просмотренного')

plt.show()
```

```python
#друг В
friends_B_forA = data[data['Subscription'] == 'A'].sort_values(by='Videos Watched')
sum_of_A = friends_B_forA['Videos Watched'].values[:200].sum()


friends_B_forB = data[data['Subscription'] == 'B'].sort_values(by='Videos Watched')
sum_of_B = friends_B_forB['Videos Watched'].values[-200:].sum()

inxA_A = friends_B_forA['Videos Watched'].index[:200]
inxA_B = friends_B_forB['Videos Watched'].index[-200:]
inxB = inxA_A + inxA_B
```

```python
plt.bar(['A', 'B'], [sum_of_A, sum_of_B])
plt.title('Вовлеченность')
plt.xlabel('Продукт')
plt.ylabel('Количество просмотренного')

plt.show()
```

```python
#добавляем новые столбцы
data['Selection Winner A'] = data.index.isin(inxA).astype(int)
data['Selection Winner B'] = data.index.isin(inxB).astype(int)
```

```python
#информация про пол для А
value_counts = data[(data['Selection Winner A'] == 1) & (data['Subscription'] == 'A')]['Gender'].value_counts()
plt.bar(value_counts.index, value_counts.values)
plt.title('Вовлеченность')
plt.xlabel('Пол')
plt.ylabel('Количество просмотренного')
plt.show()
```
